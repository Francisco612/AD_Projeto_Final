## 4.1 Replay do Fluxo: Welford, Janela Deslizante (60 min) e EWMA

- **Algoritmo de Welford:** Atualiza a média ($\mu_n$) e a soma dos quadrados dos desvios ($M_{2,n}$) em complexidade $O(1)$ por leitura, evitando erros de cancelamento numérico sem guardar o histórico todo em memória.
- **Janela Deslizante ($W = 60\text{ min}$):** Fila circular (`deque(maxlen=60)`) que soma o minuto entrante e subtrai o minuto que saiu há 60 minutos.
- **EWMA:** Aplica decaimento exponencial $s_t = \alpha x_t + (1 - \alpha) s_{t-1}$ com $\alpha = \frac{2}{W + 1} \approx 0{,}0328$, reagindo mais depressa a rampas súbitas de potência do que a média simples de 60 minutos.

```python
df_stream, df_15m, df_drifts, df_alarmes, resumo_welford = processar_fluxo_tempo_real(
    df_min,
    janela_min=60,
    limiar_z=3.5,
    delta_adwin=0.002,
)

print(f"Leituras processadas elemento a elemento: {resumo_welford['n_leituras']:,} (em {resumo_welford['tempo_s']:.2f} s)")
print("\n=== Validação Numérica: Algoritmo de Welford (Online) vs. Batch ===")
print(f"Média Global     -> Welford: {resumo_welford['mean_welford']:.8f} kW | Batch: {resumo_welford['mean_batch']:.8f} kW | Erro abs: {resumo_welford['diff_mean']:.2e}")
print(f"Variância Global -> Welford: {resumo_welford['var_welford']:.8f}    | Batch: {resumo_welford['var_batch']:.8f}    | Erro abs: {resumo_welford['diff_var']:.2e}")

plot_comparacao_dia_4_1(df_stream, dia_str=None, alpha_ewma=resumo_welford["alpha_ewma"])
```

## 4.2 Concept Drift (ADWIN) e Alarme de Anomalias

### Como distinguir *Drift* Recorrente (Sazonalidade) de *Drift* Estrutural?
1. **Drift Recorrente (Sazonalidade):** O detetor `ADWIN` assinala mudanças sempre que a média do consumo sobe na entrada do outono/inverno (ligação do aquecimento e termoacumulador) ou desce na primavera/verão e períodos de férias. Reconhece-se porque é **cíclico e reversível** (repete-se nos mesmos meses do ciclo anual) e desaparece se aplicarmos o `ADWIN` sobre os **resíduos dessazonalizados** (subtraindo o perfil esperado para aquele mês e hora do dia).
2. **Drift Estrutural (Permanente):** Corresponde a uma alteração irreversível no regime base da habitação (ex.: aquisição de veículo elétrico, instalação de bomba de calor, alteração do número de residentes ou degradação de um equipamento). Distingue-se porque **persiste mesmo após descontar a sazonalidade**, não reverte na estação seguinte e altera de forma duradoura o patamar mínimo noturno (`P10`).

```python
print(f"Total de mudanças de regime (drifts) detetadas pelo ADWIN (blocos 15 min): {len(df_drifts)}")
plot_drift_adwin_4_2(df_15m, df_drifts)

print(f"\nTotal de eventos de alarme disparados (|Z| > 3.5, janela 60 min): {len(df_alarmes)}")
df_5_alarmes = inspecionar_5_alarmes(df_min, df_alarmes, df_stream)
display(df_5_alarmes)
```

### Avaliação Qualitativa dos 5 Alarmes e Recomendação Operacional

1. **Avaliação Qualitativa dos 5 Alarmes (Reais vs. Falsos):**
   - **Falsos Alarmes (Falsos Positivos por Variância Colapsada):** Quando a habitação está em repouso (madrugada ou ausência), a média da janela de 60 minutos mantém-se muito baixa ($\approx 0{,}15\text{–}0{,}25\text{ kW}$) e o desvio padrão local colapsa ($\sigma < 0{,}05\text{ kW}$). O arranque normal do compressor do frigorífico ou do termoacumulador ($\approx 0{,}9\text{–}1{,}5\text{ kW}$) produz um $Z$-score artificialmente elevado ($|Z| > 3{,}5$), gerando um falso alarme apesar de ser operação doméstica normal.
   - **Alarmes Reais:** Quando a potência dispara para patamares elevados ($> 3{,}35\text{ kW}$, acima do limiar IQR da Parte I) sobre uma janela já ativa, acompanhados de consumo simultâneo nos sub-contadores (`Sub_1`, `Sub_2`, `Sub_3`) e ligeira queda de tensão na rede, trata-se de uma sobrecarga ou pico real de utilização relevante para monitorização.

2. **Recomendação Operacional à Sala de Controlo (Custo de Falsos Alarmes vs. Anomalias Perdidas):**
   - **Trade-off de Custo:** Um limiar baixo ($|Z| \le 3{,}0$) ou janela curta ($W \le 30\text{ min}$) provoca **fadiga de alarmes** nos operadores da sala de controlo e envia notificações inúteis aos clientes sempre que ligam um eletrodoméstico comum. Por outro lado, um limiar excessivo ($|Z| \ge 5{,}0$) deixa passar **sobrecargas prolongadas ou falhas de equipamentos** (*anomalias perdidas*).
   - **Parametrização Recomendada:**
     1. **Janela Deslizante de $W = 120\text{ minutos}$ (2 horas):** Abrange 1 a 2 ciclos completos do frigorífico e termoacumulador, estabilizando a variância local durante a noite.
     2. **Limiar de $|Z| > 3{,}5$ com piso de desvio padrão ($\sigma_{\min} = 0{,}25\text{ kW}$):** Impede que pequenas oscilações de repouso disparem alarmes falsos.
     3. **Regra de Persistência ($\ge 3\text{ minutos consecutivos}$ acima do limiar):** Filtra picos transitórios de arranque de motores (1 minuto) e garante que apenas anomalias sustentadas chegam ao operador.
